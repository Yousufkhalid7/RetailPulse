##Data-card Expectations
- 500K Rows, 8 Columns
- Negative Quantity values = cancelled transactions
- 'C' in transaction id = cancelled transactions
- Price should be > 0
- Every row should have a customer number

Below, I will test these expectations with actual data

```python
import pandas as pd
df = pd.read_csv('../data/sales transaction v.4a.csv')
print(df.shape)
print(df.head(10))
```

```python
#Date month/day/year or day/month/year?
parts = df["Date"].str.split("/", expand=True).astype(int)
print("first part max:", parts[0].max(), "| second part max:", parts[1].max())
```

```python
#real-time range?
df["Date"] = pd.to_datetime(df["Date"], format="%m/%d/%Y")
print(df["Date"].min(), df["Date"].max())
```

```python
is_cancel = df["TransactionNo"].str.startswith("C")
neg_qty = df["Quantity"] < 0
pd.crosstab(is_cancel, neg_qty)
```
